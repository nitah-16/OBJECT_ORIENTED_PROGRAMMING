QUESTION: Write a python program that creates a student class ccontaining a student's name and age. Add a method called introduce()that displays the student's information

```python
#first define the student class
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
#method to introduce the student
    def introduce(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")
#then create the student object
student1 = Student("Anita", 21)
#Then call the introduce()method using the object
student1.introduce()
```

QUESTION 
Write a python program that creates a phone class each phone should have a brand and storage capacity. Add methods that dispaly phone information and stimulate making a call. Create two phone objects and call their methods.



```python
class Phone:
    #we initialize
    def __init__(self, brand, storage_capacity):
        #then we store
        self.brand = brand
        self.storage_capacity = storage_capacity
#then we display
    def display_info(self):
        print("Brand: " + self.brand)
        print("Storage Capacity: " + str(self.storage_capacity) + " GB")

    def make_call(self, number):
        print(f"{self.brand} is calling {number}...")


# Create two Phone objects
phone1 = Phone("Samsung", 128)
phone2 = Phone("iPhone", 256)

# Call methods for phone1
print("Phone 1:")
phone1.display_info()
phone1.make_call("0700123456")

print()

# Call methods for phone2
print("Phone 2:")
phone2.display_info()
phone2.make_call("0750987654")
```

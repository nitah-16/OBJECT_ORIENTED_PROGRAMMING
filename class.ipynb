```python
class HostelRoom:
    def __init__(self, room_number: str, capacity: int):
        self.room_number = room_number          # Public state
        self._capacity = capacity               # Internal state (protected)
        self._current_occupants = 0             # Internal state (protected)

    def get_capacity(self) -> int:
        return self._capacity

    def get_occupants(self) -> int:
        return self._current_occupants

    def add_occupant(self) -> bool:
        if self._current_occupants >= self._capacity:
            print(f"Rejected: Cannot add occupant. Room {self.room_number} is full.")
            return False
        self._current_occupants += 1
        print(f"Success: Added 1 occupant. Total: {self._current_occupants}/{self._capacity}")
        return True

    def remove_occupant(self) -> bool:
        if self._current_occupants <= 0:
            print(f"Rejected: Cannot remove occupant. Room {self.room_number} is empty.")
            return False
        self._current_occupants -= 1
        print(f"Success: Removed 1 occupant. Total: {self._current_occupants}/{self._capacity}")
        return True


# --- Demonstration ---
if __name__ == "__main__":
    # Create a room with a capacity of 2 students
    print("--- Creating Room ---")
    room = HostelRoom(room_number="101A", capacity=2)
    print(f"Room Created: Capacity = {room.get_capacity()}\n")

    # 1. Attempt to add three students
    print("--- 1. Adding 3 Students ---")
    room.add_occupant()
    room.add_occupant()
    room.add_occupant()  # This should be rejected (capacity is 2)
    print()

    # 2. Remove students until the room is empty
    print("--- 2. Removing Students Until Empty ---")
    room.remove_occupant()
    room.remove_occupant()
    print()

    # 3. Attempt one more removal
    print("--- 3. Attempting One More Removal ---")
    room.remove_occupant()  # This should be rejected (already empty)
```

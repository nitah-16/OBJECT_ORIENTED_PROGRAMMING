Currently, our system uses procedural programming where student data is stored as raw dictionary structures while separate global functions (calculate_average, assign_grade, determine_status) operate on that data from the outside.

In Python, Object-Oriented Programming (OOP) allows us to unify this using a class Student. By encapsulating attributes (self.name, self.courses) and methods (calculate_averages(), get_status()) inside a single class template, each student object manages its own state and behavior. Instead of passing nested dictionaries to external functions, we simply call student.get_status(). This prevents data corruption, improves code reusability, and makes complex systems far easier to scale.

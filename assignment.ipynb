```python
class BodaBoda:
    # 2. Represents the company appropriately (Class Attribute)
    company_name = "Mukono Campus Rides"

    # 3. Uses __init__() to initialize the information for each boda-boda
    def __init__(self, reg_number, rider_name, current_location):
        self.reg_number = reg_number
        self.rider_name = rider_name
        self.current_location = current_location
        self.completed_trips = 0  # Starts at 0 for every new boda-boda

    # 4. Method to complete a trip
    def complete_trip(self):
        self.completed_trips += 1
        print(f"{self.rider_name} ({self.reg_number}) completed a trip.")

    # 4. Method to change current location
    def change_location(self, new_location):
        self.current_location = new_location
        print(
            f"{self.rider_name} ({self.reg_number}) relocated to {new_location}."
        )

    # Method to display current information
    def display_info(self):
        print("\n--- Boda-Boda Details ---")
        print(f"Company: {BodaBoda.company_name}")
        print(f"Registration Number: {self.reg_number}")
        print(f"Rider Name: {self.rider_name}")
        print(f"Current Location: {self.current_location}")
        print(f"Completed Trips: {self.completed_trips}")


# --- DEMONSTRATION OF PROGRAM EXECUTION ---

# 5. Creates two different objects
boda1 = BodaBoda("UEB 123A", "Kizito", "Main Gate")
boda2 = BodaBoda("UED 456B", "Muwanguzi", "Bishop Tucker")

# 6. Demonstration steps:
# o The first boda-boda completes two trips
boda1.complete_trip()
boda1.complete_trip()

# o The second boda-boda completes one trip
boda2.complete_trip()

# o The first boda-boda changes location
boda1.change_location("Tech Park")

# o The final details of both objects are displayed
boda1.display_info()
boda2.display_info()
```
